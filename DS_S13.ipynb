```python
data = {
    'Patient_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Ravi', 'Sneha', 'Arjun', 'Priya', 'Karan', 'Meena', 'Vikas', 'Neha', 'Amit', 'Ritu'],
    'Age': [25, 47, 33, 29, 52, 61, 40, 36, 55, 30],
    'Gender': ['M', 'F', 'M', 'F', 'M', 'F', 'M', 'F', 'M', 'F'],
    'Disease': ['Diabetes', 'Hypertension', 'Flu', 'Asthma', 'Arthritis', 'Diabetes', 'Migraine', 'Flu', 'Hypertension', 'Asthma'],
    'Blood_Pressure': [50, 150, 120, 125, 140, 160, 110, 115, 155, 250],
    'Heart_Rate': [85, 90, 76, 88, 95, 102, 70, 80, 98, 84],
    'Cholesterol': [200, 240, 180, 190, 230, 250, 175, 185, 245, 195],
    'Smoking': ['No', 'Yes', 'No', 'No', 'Yes', 'Yes', 'No', 'No', 'Yes', 'No'],
    'Treatment_Cost': [5000, 12000, 3000, 7000, 15000, 18000, 4000, 3500, 16000, 6500]
}
df=pd.DataFrame(data)
df.head()
```
