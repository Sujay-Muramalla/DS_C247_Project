```python
import pandas as pd
```

```python
Region_A = pd.Series ([100,200,132,40,5], index=["Jan","Feb","Mar","Apr","May"])
Region_B = pd.Series ([60,7,88,99,150], index=["Jan","Feb","Mar","Apr","May"])

```

```python
def sales_category(sales):
    if sales < 100:
        return "Low"
    elif sales < 5000:
        return "Medium"
    else:
        return "High"
    
tagged_region_A = Region_A.apply(sales_category)
print (tagged_region_A)
```
