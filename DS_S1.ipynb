```python
# EXERCISE 14: Fenced Matrix (All boundary elements 1, rest all zero)
rows, cols = 3, 4

# Create a matrix filled with zeros
matrix = np.zeros((rows, cols), dtype=int)

# Set all boundary elements to 1
matrix[0, :] = 1      # First row
matrix[-1, :] = 1     # Last row
matrix[:, 0] = 1      # First column
matrix[:, -1] = 1     # Last column

print(f"Fenced Matrix ({rows}x{cols}):")
print(matrix)
```

```python
# EXERCISE 14: Alternative Approaches

# Approach 2: Using np.pad()
print("Approach 2: Using np.pad()")
rows, cols = 3, 4
inner_matrix = np.zeros((rows-2, cols-2), dtype=int)
matrix2 = np.pad(inner_matrix, 1, constant_values=1)
print(matrix2)

# Approach 3: Using np.ones() and setting interior to 0
print("\nApproach 3: Using np.ones() and setting interior to 0")
matrix3 = np.ones((rows, cols), dtype=int)
matrix3[1:-1, 1:-1] = 0
print(matrix3)

# Approach 4: Using list comprehension
print("\nApproach 4: Using list comprehension")
matrix4 = np.array([[1 if i == 0 or i == rows-1 or j == 0 or j == cols-1 else 0 
                     for j in range(cols)] for i in range(rows)])
print(matrix4)
```
