```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Smoking": ["Yes", "No", "Yes", "No", "Yes", "Yes", "No"]
}

df = pd.DataFrame(data)

plt.figure(figsize=(4,2))
plt.bar(df["Smoking"].value_counts().index,
        df["Smoking"].value_counts().values)
plt.xlabel("Smoking - Yes or No")
plt.ylabel("Count of Patients")
plt.show()
```
