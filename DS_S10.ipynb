```python
import numpy as np
import pandas as pd
data = {
    'Patient_ID': ['P001', 'P002', 'P003', 'P004', 'P005'],
    'Heart_Rate (bpm)': [78, np.nan, 85, 72, 90],
    'Blood_Pressure (mmHg)': ['120/80', '118/76', np.nan, '110/70', '130/85'],
    'Temperature (°C)': [36.8, 37.2, 38.0, np.nan, 37.5],
    'Oxygen_Saturation (%)': [98, 96, 94, 97, np.nan]
}

df = pd.DataFrame(data)
df.isnull().sum()
100*df.isna().mean()
```

```python
#ignore the columns which have more than 50% missing values
#otherwise we impute
#median is preferred when compared with mean because mean is affected by outliers but median is not affected by outliers

heart_rate_mean = df['Heart_Rate (bpm)'].mean()
print(heart_rate_mean)
```

```python
df['Heart_Rate (bpm)'] = df['Heart_Rate (bpm)'].fillna(heart_rate_mean)
```

```python
df['Heart_Rate (bpm)']
```

```python
temp_median = df['Temperature (°C)'].median()
print(temp_median)
df['Temperature (°C)'] = df['Temperature (°C)'].fillna(temp_median)
```

```python
df['Temperature (°C)']
```
